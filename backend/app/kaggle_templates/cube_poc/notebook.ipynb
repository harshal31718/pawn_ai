```python
# PAWN cube proof-of-transport kernel.
# PAWN replaces __PAWN_PAYLOAD_B64__ with base64(json) before each push.
# The prompt/input is base64-injected (never string-interpolated) so kernel
# source can't be broken or injected via user input.
import base64, json, os

payload = json.loads(base64.b64decode("__PAWN_PAYLOAD_B64__").decode())

def findCube(n):
    return n ** 3

n = int(payload["input"])
result = {"input": n, "result": findCube(n)}

os.makedirs("/kaggle/working", exist_ok=True)
with open("/kaggle/working/out.json", "w") as f:
    json.dump(result, f)

print(result)
```
